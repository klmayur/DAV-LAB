```python
import numpy as np

arr = np.array([35, 82, 64, 91, 48])

print("array:", arr)
print("mean:", arr.mean())
print("reshaped 2D:\n", arr.reshape(5, 1))
print("sliced(index 1 to 3):", arr[1:4])


import pandas as pd

data = {
    "Name": ['rahul', 'megha', 'kiran', 'pooja'],
    "Marks": [68, 85, 92, 74],
    "Branch": ['ISE', 'CSE', 'ECE', 'CSE'],
}

df = pd.DataFrame(data)

print("\nFirst two rows:\n", df.iloc[0:2])

print("\nMarks column\n", df['Marks'])

print("\nRows where Marks > 70:\n", df[df['Marks'] > 70])
```
